## 1. Probability

### 1.1 Definition

Defined as the mathematical language for quantifying uncertainity.

### 1.2 Sample Spaces and Events

**Sample space** $(\Omega)$: the set of all possible outcomes of an experiment<br>
**Sample outcomes** / **Elements** / **Realisations**: $\omega \in \Omega$<br>
**Events (A)** : subsets of $\Omega$ / $A \subseteq \Omega$

$A^c = $ not $A$ or **complement** of $A \equiv \{ \omega \in \Omega : \omega \notin A \} $ <br>
$\Omega^c = \emptyset$

The **union** of events $A$ and $B$ is defined as:
$$ A \cup B = \{ \omega \in \Omega : \omega \in A \text{ or } \omega \in B \text{ or } \omega \in \text{both } \} $$

If $A_1, A_2, \dotsc $ is a sequence of sets, then
$$ \bigcup_{i=1}^\infty A_i = \bigl \{ \omega \in \Omega : \omega \in A_i \; \text{ for at least one } i \bigr \}$$

The **intersection** of $A$ and $B$ is defined as:
$$A \cap B = \{ \omega \in \Omega : \omega \in A \text{ and } \omega \in B \}$$

If $A_1, A_2, \dotsc $ is a sequence of sets, then
$$ \bigcap_{i=1}^\infty A_i = \bigl \{ \omega \in \Omega : \omega \in A_i \; \text{ for all } i \bigr \}$$

**Set difference** $(A - B) = \left\{ \omega \in \Omega : \omega \in A ,\; \omega \notin B \right\}$.<br>
If $A$ is a finite set, let $|A|$ denote the number of elements in $A$.

| notation           | meaning                                            |
|--------------------|----------------------------------------------------|
| $\Omega$           | sample space                                       |
| $\omega$           | sample outcome                                     | 
| $A$                | event (subset of $\Omega$)                         |
| $\vert A \vert$    | number of elements in $A$ (if finite)              |
| $A^c$              | complement of $A$ (not $A$)                        |
| $A \cup B$         | union ($A$ or $B$)                                 |
| $A \cap B$ or $AB$ | intersection ($A$ and $B$)                         |
| $A - B$            | set difference (points in $A$ but not in $B$)      |
| $A \subset B$      | set inclusion ($A$ is a subset of or equal to $B$) |
| $\emptyset$        | null event (always false)                          |
| $\Omega$           | true event (always true)                           |

We say that $A_1, A_2, \dotsc $ are **disjoint** or **mutually exclusive** if $A_i \cap A_j = \emptyset$ whenever $i \neq j$.

A **partition** of $\Omega$ is a sequence of **disjoint** sets $A_1, A_2, \dotsc $ such that $\bigcup_{i=1}^\infty A_i = \Omega$.

Given an event $A$, the **indicator function of $A$** is defined as:

$$ I_A(\omega) = I(\omega \in A) = \begin{cases}
1 &\text{if } \omega \in A \\
0 &\text{if } \omega \notin A
\end{cases}$$

Suppose we have a sequence of sets $A_1, A_2, \dotsc $. These sets are:
1. **Monotone increasing** $\phantom{\:} \mathrm{if} \;\; A_1 \subset A_2 \subset \dotsb \quad \& \quad \lim_{n \to \infty} A_n = \bigcup_{i=1}^\infty A_i $.
2.  **Monotone decreasing** $ \mathrm{if} \;\; A_1 \supset A_2 \supset \dotsb \quad \& \quad \lim_{n \to \infty} A_n = \bigcap_{i=1}^n A_i$.

In either case, we will write $A_n \rightarrow A$.

### 1.3 Probability

A function $\mathbb{P}$ that assigns a real number $\mathbb{P}(A)$ to each event $A$.<br>
$\mathbb{P}$ is a **probability distribution** / **probability measure** that must satisfy three axioms:

- **Axiom 1**: $\mathbb{P}(A) \geq 0 \; , \; \forall A$
- **Axiom 2**: $\mathbb{P}(\Omega) = 1$
- **Axiom 3**: If $A_1, A_2, \dotsc $ are disjoint then

$$ \mathbb{P} \left( \bigcup_{i=1}^\infty A_i \right) = \sum_{i=1}^\infty \mathbb{P}(A_i) $$

2 Common interpretations of $\mathbb{P}(A)$:
1. **Frequentist Interpretation:** in an infinitely long, unpredictable sequence, $\mathbb{P}(A)$ is defined as the long-run relative frequency of Event $A$ occuring.
2. **Bayesian Interpretation:** $\mathbb{P}(A)$ measures an observer's degree of belief / confidence that $A$ is true, given the available evidence.

Properties that can be derived from the 3 axioms:
- $\mathbb{P}(\emptyset) = 0$
- $A \subset B \implies \mathbb{P}(A) \leq \mathbb{P}(B)$
- $0 \leq \mathbb{P}(A) \leq 1 $
- $\mathbb{P}(A^c) = 1 - \mathbb{P}(A)$
- $A \cap B = \emptyset \implies \mathbb{P}(A \cup B) = \mathbb{P}(A) + \mathbb{P}(B)$

**Theorem (Continuity of Probabilities)**.  If $A_n \rightarrow A$ then $\mathbb{P}(A_n) \rightarrow \mathbb{P}(A)$ as $n \rightarrow \infty$.

**Proof**. Suppose that $A_n$ is monotone increasing s.t. $A_1 \subset A_2 \subset \dotsb $<br>
Let $A = \lim_{n \to \infty} A_n = \bigcup_{i=1}^\infty A_i $

Define $B_1 = A_1$ and $B_{n+1} = A_{n+1} - A_n$ for $n > 1$.

$B_1, B_2, \dotsc $ are disjoint by construction as $B_i \cap B_j = \emptyset, \text{ for } \; i \neq j $.<br>
This also means that $A_n = \bigcup_{i=1}^n A_i = \bigcup_{i=1}^n B_i$ for all $n$.

From **Axiom 3**,

$$ \mathbb{P}(A_n) = \mathbb{P} \left ( \bigcup_{i=1}^n B_i \right )  = \sum_{i=1}^n \mathbb{P}(B_i) $$
and hence, using **Axiom 3** again:
$$ \lim_{n \to \infty} \mathbb{P}(A_n) = \lim_{n \to \infty} \sum_{i=1}^n \mathbb{P}(B_i) = \sum_{i=1}^\infty \mathbb{P}(B_i) = \mathbb{P}\left( \bigcup_{i=1}^\infty B_i \right) = \mathbb{P}(A) $$

### 1.4 Probability on Finite Sample Spaces

$|A|$ denotes the number of elements in $A$ / the **cardinality** of the set.

If $\Omega$ is finite and each outcome is equally likely, then $ \mathbb{P}(A) = \frac{|A|}{|\Omega|} $ which is called the **uniform probability distribution**.

To compute probabilities, we need to count the number of points in an Event $A$.<br>
Methods for counting points are called **combinatorial methods**.

1. Given $n$ objects, the number of ways of ordering these objects is $n!$

2. The number of **permutations** of a subset of size $k$, chosen from a set of $n$ distinct objects is
$$ \text{nPk} = \frac{n!}{(n-k)!} $$

3. The number of **combinations** of a subset of size $k$, chosen from a set of $n$ distinct objects (where order does not matter) is

$$ \text{nCk} = \binom{n}{k} =  \frac{n!}{k!(n-k)!} $$

**Note:**
$$ \binom{n}{0} = \binom{n}{n} = 1 \text{ and } \binom{n}{k} = \binom{n}{n - k} $$

### 1.5 Independent Events

Two events $A$ and $B$ are **independent** if
$ \mathbb{P}(AB) = \mathbb{P}(A) \mathbb{P}(B) $
and we write $A \perp \!\! \perp B$.

A set of events $\{ A_i : i \in I \}$ is independent if

$$ \mathbb{P} \left( \bigcap_{i \in J} A_i \right) = \prod_{i \in J} \mathbb{P}(A_i) $$

for every finite subset $J$ of $I$ $ \; (J \subset I) $.

**Some key points:**
- Independence is sometimes assumed and sometimes derived.
- Disjoint events with positive probability are not independent.

### 1.6 Conditional Probability

Asssuming $ \mathbb{P}(B) > 0 $ then the **conditional probability** of $A$ given $B$ is defined as
$$ \mathbb{P}(A | B) = \frac{\mathbb{P}(AB)}{\mathbb{P}(B)} $$

$ \mathbb{P}(A | B) $ can be thought of as the proportion of times $A$ occurs given that $B$ has already occcured.

For for fixed $B$ such that $\mathbb{P}(B) > 0 \: , \;\;  \mathbb{P}(\cdot | B)$ is a probability (that satisfies the axioms of probability): 
- $\mathbb{P}(A | B) \geq 0 $
- $\mathbb{P}(\Omega | B) = 1 $
- If $A_1, A_2, \dotsc $ are disjoint then
$$ \mathbb{P} \left( \bigcup_{i = 1}^\infty A_i | B \right) = \sum_{i = 1}^\infty \mathbb{P}(A_i | B) $$

In general, $\mathbb{P}(A | \cdot)$ does **not** satisfy the axioms of probability for fixed $A$.

**Lemma**. $A$ and $B$ are independent if and only if: $ \; \mathbb{P}(A | B) = \mathbb{P}(A) \; \text{ and } \; \mathbb{P}(B | A) = \mathbb{P}(B)$.

Also, for any pair of events $A$ and $B$:

$$ \mathbb{P}(AB) = \mathbb{P}(A|B) \cdot \mathbb{P}(B) = \mathbb{P}(B|A) \cdot \mathbb{P}(A) $$

*Note: this is the Conditional Probability formula rearranged*

### 1.7 Bayes' Theorem

**Theorem (The Law of Total Probability)**.  Let $A_1, \dotsc, A_k$ be a partition of $\Omega$. <br>
Then, for any event $B$,

$$ \mathbb{P}(B) = \sum_{i=1}^k \mathbb{P}(B | A_i) \cdot \mathbb{P}(A_i) $$

**Proof**.  Let $C_j = BA_j$. Also note that $C_1, \dotsc C_k \; $ are disjoint and that $B = \bigcup_{i=1}^k C_j$. <br>
Hence:

$$ \mathbb{P}(B) = \mathbb{P} \left( \bigcup_{j} C_j \right) = \sum_j \mathbb{P}(C_j)  = \sum_j \mathbb{P}(BA_j) = \sum_j \mathbb{P}(B | A_j) \cdot \mathbb{P}(A_j) $$

**Theorem (Bayes' Theorem)**.  Let $A_1, \dotsc, A_k \;$ be a partition of $\Omega \:$ such that $\mathbb{P}(A_i) > 0$ for each $i$. <br>
If $\mathbb{P}(B) > 0$, then for each $i = 1, \dotsc, k$

$$ \mathbb{P}(A_i | B) = \frac{\mathbb{P}(A_i \cap B)}{\mathbb{P}(B)} = \frac{\phantom{\sum_j} \mathbb{P}(B | A_i) \cdot \mathbb{P}(A_i)}{\sum_j \mathbb{P}(B | A_j) \cdot \mathbb{P}(A_j)} $$

### 1.9 Technical Appendix

Generally, it is not feasible to assign probabilities to all the subsets of a sample space $\Omega$.  Instead, one restricts attention to a set of events called a **$\sigma$-algebra** or a **$\sigma$-field**.

It is essentially a sophisticated way of stating that if we assign a probability to certain events $(\text{e.g. } \; \mathbb{P}(A), \; \mathbb{P}(B))$ we also need to assign probabilities to their **complements** $(\mathbb{P}(A^c), \; \mathbb{P}(B^c))$ and to their **union** $(\mathbb{P}(A \cup B))$

A **$\sigma$-algebra** $(\mathcal{A})$ is a collection of sets in $\Omega$ such that:
- $\emptyset \in \mathcal{A}\: , \;\; \Omega \in \mathcal{A}$
- If $A_1, A_2, \dotsc \in \mathcal{A}$, then $\bigcup_{i=1}^\infty A_i \in \mathcal{A}$
- $A \in \mathcal{A} \implies A^c \in \mathcal{A}$

A $\sigma$-algebra on a set $X$, is a non-empty collection of subsets of $X$ that is **closed** under complement, countable unions and countable intersections.

To be **closed** under certain operations (e.g. complement, union, intersection etc.) means that if you apply those operations to elements (i.e. subsets) within the $\sigma$-algebra, the resulting subset is also guaranteed to be in the $\sigma$-algebra.

A **probability space**  $(\Omega, \mathcal{A}, \mathbb{P})$ is a mathematical framework to represent and analyse / model uncertain phenomena of interest as an **experiment** with several (possibly infinite) mutually exclusive **outcomes**. Its components include:
- **Sample space** $(\Omega)$: contains all possible outcomes of an experiment
- **$\sigma$-algebra** $(\mathcal{A})$: a collection of subsets of $\Omega$, including $\Omega$ itself
- **Probability measure** $(\mathbb{P})$: a function $\mathbb{P} : \mathcal{A} \rightarrow [0,1]$ that assigns probabilities to the events in $\mathcal{A}$ and satisfies the 3 axioms of probability
