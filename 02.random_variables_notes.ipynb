## 2. Random Variables

### 2.1 Introduction

A **random variable** is a mapping $ X: \omega \rightarrow \mathbb{R} $ that assigns a real number $X(\omega)$ to each outcome $\omega$.

A **mapping** is a rule that takes a number as an **input** and transforms it into another number as an **output**.<br>
The set of numbers / inputs you start with is called the **domain**.<br>
The set of numbers they become (after a mapping) is called the **range** (or **image**).<br>
The **preimage** of a value is the set of input value(s) that produce it when put into a function.

A **function** is a mapping that assigns each input in the domain to exactly one output in the range.

Types of mappings:
- **One-to-one (Injective):** each value in the range corresponds to one value in the domain (e.g. $y=x$).
- **Many-to-one:** if some values in the range correspond to more than one value in the domain (e.g. $y=x^2$).
- **One-to-many:** values in the domain can map to more than 1 value in the range (e.g. $y=\pm\sqrt{x}$).<br> **Note:** not a function as input values don't map to exactly one output value. 

Given a random variable $X$ and a subset of the real line $A$, we define the **preimage** of $A$ (under $X$) as<br>
$$ X^{-1}(A) = \{ \omega \in \Omega : X(\omega) \in A \} $$

and let

$$
\begin{align}
\mathbb{P}(X \in A) = \mathbb{P}(X^{-1}(A)) &= \mathbb{P}(\{ \omega \in \Omega : X(\omega) \in A \}) \\
\mathbb{P}(X = x) = \mathbb{P}(X^{-1}(x)) &= \mathbb{P}(\{ \omega \in \Omega : X(\omega) = x \} )
\end{align}
$$

where $X$ denotes a random variable and $x$ denotes a possible value of $X$.

### 2.2 Distribution Functions and Probability Functions

The **cumulative distribution function** (or **CDF**) is the function $F_{x} : \mathbb{R} \rightarrow [0,1] $

$$ F_{X}(x) = \mathbb{P}(X \leq x) $$

**Theorem 2.7**. Let $X$ have CDF $F$ and let $Y$ have CDF $G$. If $F(x) = G(x)$ for all $x$ then $\mathbb{P}(X \in A) = \mathbb{P}(Y \in A)$ for all $A$.

**Theorem 2.8**. A function $F$ mapping the real line to $[0,1]$ is a CDF for some probability iff $F$ satisfies the following conditions:
- $F$ is **non-decreasing**: $x_1 < x_2 \implies F(x_1) \leq F(x_2)$
- $F$ is **normalised**: $\lim_{n \rightarrow -\infty} F(x) = 0$ and $\lim_{n \rightarrow +\infty} F(x) = 1$.
- $F$ is **right continuous**: i.e. $F(x) = F(x^+)$ for all $x$, where
$$ F(x^+) = \lim_{\substack{y \rightarrow x \\ y > x}} F(y) $$
-  $F$ is defined for all values of $x$

A function is **left continuous** at $a$ if:
$$ \underbrace{\lim_{x \to a^{-}} f(x)}_{\text{left-hand limit at } a} = \quad f(a) $$

A function is **right continuous** at $a$ if:
$$ \underbrace{\lim_{x \to a^{+}} f(x)}_{\text{right-hand limit at } a} = \quad f(a) $$

**Proof**. Suppose that $F$ is a **CDF**.  Let us show that $F$ is **right continuous**.<br>
Let $x$ be a real number and let $ y_1, y_2, \dotsc $ be a sequence of real numbers such that $ y_1 > y_2 > \dotsb $ and $ \lim_i y_i = x$.<br>
Let $ A_i = (-\infty, y_i] $ and let $ A = (-\infty, x] $.

Note that $ A = \bigcap_{i=1}^\infty A_i$ and also note that $A_1 \supset A_2 \supset \dotsb $ is monotone decreasing. <br>
From these properties we can derive $\lim_i \mathbb{P}(A_i) = \mathbb{P}(\bigcap_i A_i)$. Thus,

$$ F(x) = \mathbb{P}(A) = \mathbb{P}(\bigcap_i A_i) = \lim_i \mathbb{P}(A_i) = \lim_i F(y_i) = F(x^+) $$

$X$ is **discrete** if it takes countably many values $ \{ x_1, x_2, \dotsc \} $

We define the **probability (mass) function** (or **PMF**) for $X$ as
$$ f_X(x) = \mathbb{P}(X = x) $$

Thus, $ f_X(x) \geq 0$ for all $ x \in \mathbb{R} $ and $ \sum_i f_X(x_i) = 1 $.

**Note:** Often we write $f_X$ and $F_X$ simply as $f$ and $F$.

The CDF of $X$ ($F_X$) is related to the PMF of $X$ ($f_X$) by:

$$ F_X(x) = \mathbb{P}(X \leq x) = \sum_{x_i \leq x} f_X(x_i) $$

$X$ is **continuous** if there exists a function $f_X$ such that $ f_X(x) \geq 0 $ for all $x$, $ \int_{-\infty}^\infty f_X(x) dx = 1 $ and for every $ a \leq b $:

$$ \mathbb{P}(a < X < b ) = \int^{b}_{a} f_X(x) dx $$

The function $f_X$ is called the **probability density function** (or **PDF**).<br>
We also have that the corresponding CDF:
$$ F_X(x) = \int_{-\infty}^x f_X(t) dt $$

and $f_X(x) = F'_X(x)$ at all points $x$ at which $F_X$ is **differentiable**.

**Note:** Sometimes we write $ \int f(x) dx \: $ or $ \: \int f $ to mean $ \int_{-\infty}^\infty f_X(x) dx $.

**Warning**: Note that if $X$ is continuous then $ \mathbb{P}(X = x) = 0 $ for every $x$.<br>
$\phantom{\text{Warning}}$ We only have $f(x) = \mathbb{P}(X = x)$ for discrete random variables.<br>
$\phantom{\text{Warning}}$ We get probabilities from a PDF by **integrating**.

**Note:** A PDF can be **bigger than $1$** in some places, unlike a PMF (e.g. $f(x)=5$ for $x \in [0, \frac{1}{5}]$ and $0$ otherwise).<br>
$\phantom{\text{Note:}}$ A PDF can also be **unbounded**, as long as it has a finite integral = 1 (e.g. $f(x)=\frac{2}{3}x^{-\frac{1}{3}}$ for $0<x<1$ and $0$ otherwise).
