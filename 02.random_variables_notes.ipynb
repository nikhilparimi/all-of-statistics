## 2. Random Variables

### 2.1 Introduction

A **random variable** is a mapping $ X: \omega \rightarrow \mathbb{R} $ that assigns a real number $X(\omega)$ to each outcome $\omega$.

A **mapping** is a rule that takes a number as an **input** and transforms it into another number as an **output**.<br>
The set of numbers / inputs you start with is called the **domain**.<br>
The set of numbers they become (after a mapping) is called the **range** (or **image**).<br>
The **preimage** of a value is the set of input value(s) that produce it when put into a function.

A **function** is a mapping that assigns each input in the domain to exactly one output in the range.

Types of mappings:
- **One-to-one (Injective):** each value in the range corresponds to one value in the domain (e.g. $y=x$).
- **Many-to-one:** if some values in the range correspond to more than one value in the domain (e.g. $y=x^2$).
- **One-to-many:** values in the domain can map to more than 1 value in the range (e.g. $y=\pm\sqrt{x}$).<br> **Note:** not a function as input values don't map to exactly one output value. 

Given a random variable $X$ and a subset of the real line $A$, we define the **preimage** of $A$ (under $X$) as<br>
$$ X^{-1}(A) = \{ \omega \in \Omega : X(\omega) \in A \} $$

and let

$$
\begin{align*}
\mathbb{P}(X \in A) = \mathbb{P}(X^{-1}(A)) &= \mathbb{P}(\{ \omega \in \Omega : X(\omega) \in A \}) \\
\mathbb{P}(X = x) = \mathbb{P}(X^{-1}(x)) &= \mathbb{P}(\{ \omega \in \Omega : X(\omega) = x \} )
\end{align*}
$$

where $X$ denotes a random variable and $x$ denotes a possible value of $X$.

### 2.2 Distribution Functions and Probability Functions

The **cumulative distribution function** (or **CDF**) is the function $F_{x} : \mathbb{R} \rightarrow [0,1] $

$$ F_{X}(x) = \mathbb{P}(X \leq x) $$

**Theorem 2.7**. Let $X$ have CDF $F$ and let $Y$ have CDF $G$. If $F(x) = G(x)$ for all $x$ then $\mathbb{P}(X \in A) = \mathbb{P}(Y \in A)$ for all $A$.

**Theorem 2.8**. A function $F$ mapping the real line to $[0,1]$ is a CDF for some probability iff $F$ satisfies the following conditions:
- $F$ is **non-decreasing**: $x_1 < x_2 \implies F(x_1) \leq F(x_2)$
- $F$ is **normalised**: $\lim_{x \rightarrow -\infty} F(x) = 0$ and $\lim_{x \rightarrow +\infty} F(x) = 1$.
- $F$ is **right continuous**: i.e. $F(x) = F(x^+)$ for all $x$, where
$$ F(x^+) = \lim_{\substack{y \rightarrow x \\ y > x}} F(y) $$
-  $F$ is defined for all values of $x$

A function is **left continuous** at $a$ if:
$$ \underbrace{\lim_{x \to a^{-}} f(x)}_{\text{left-hand limit at } a} = \quad f(a) $$

A function is **right continuous** at $a$ if:
$$ \underbrace{\lim_{x \to a^{+}} f(x)}_{\text{right-hand limit at } a} = \quad f(a) $$

**Proof**. Suppose that $F$ is a **CDF**.  Let us show that $F$ is **right continuous**.<br>
Let $x$ be a real number and let $ y_1, y_2, \dotsc $ be a sequence of real numbers such that $ y_1 > y_2 > \dotsb $ and $ \lim_i y_i = x$.<br>
Let $ A_i = (-\infty, y_i] $ and let $ A = (-\infty, x] $.

Note that $ A = \bigcap_{i=1}^\infty A_i$ and also note that $A_1 \supset A_2 \supset \dotsb $ is monotone decreasing. <br>
From these properties we can derive $\lim_i \mathbb{P}(A_i) = \mathbb{P} \left ( \bigcap_i A_i \right ) $. Thus,

$$ F(x) = \mathbb{P}(A) = \mathbb{P} \biggl ( \bigcap_i A_i \biggr ) = \lim_i \mathbb{P}(A_i) = \lim_i F(y_i) = F(x^+) $$

$X$ is **discrete** if it takes countably many values $ \{ x_1, x_2, \dotsc \} $

We define the **probability (mass) function** (or **PMF**) for $X$ as
$$ f_X(x) = \mathbb{P}(X = x) $$

Thus, $ f_X(x) \geq 0$ for all $ x \in \mathbb{R} $ and $ \sum_i f_X(x_i) = 1 $.

**Note:** Often we write $f_X$ and $F_X$ simply as $f$ and $F$.

The CDF of $X$ ($F_X$) is related to the PMF of $X$ ($f_X$) by:

$$ F_X(x) = \mathbb{P}(X \leq x) = \sum_{x_i \leq x} f_X(x_i) $$

$X$ is **continuous** if there exists a function $f_X$ such that $ f_X(x) \geq 0 $ for all $x$, $ \int_{-\infty}^\infty f_X(x) \ dx = 1 $ and for every $ a \leq b $:

$$ \mathbb{P}(a < X < b ) = \int^{b}_{a} f_X(x) \ dx $$

The function $f_X$ is called the **probability density function** (or **PDF**).<br>
We also have that the corresponding CDF:
$$ F_X(x) = \int_{-\infty}^x f_X(t) \ dt $$

and $f_X(x) = F'_X(x)$ at all points $x$ at which $F_X$ is **differentiable**.

**Note:** Sometimes we write $ \int f(x) \ dx \: $ or $ \: \int f $ to mean $ \int_{-\infty}^\infty f_X(x) \ dx $.

**Warning**: Note that if $X$ is continuous then $ \mathbb{P}(X = x) = 0 $ for every $x$.<br>
$\phantom{\text{Warning}}$ We only have $f(x) = \mathbb{P}(X = x)$ for discrete random variables.<br>
$\phantom{\text{Warning}}$ We get probabilities from a PDF by **integrating**.

**Note:** A PDF can be **bigger than $1$** in some places, unlike a PMF (e.g. $f(x)=5$ for $x \in [0, \frac{1}{5}]$ and $0$ otherwise).<br>
$\phantom{\text{Note:}}$ A PDF can also be **unbounded**, as long as it has a finite integral = 1 (e.g. $f(x)=\frac{2}{3}x^{-\frac{1}{3}}$ for $0<x<1$ and $0$ otherwise).

**Lemma**. Let $F$ be the CDF for a random variable $X$. Then:
1. $\mathbb{P}(X = x) = F(x) - F(x^-) = P(X \leq x) - P(X < x)$<br> Where $\underbrace{F(x^-)}_{\substack{\text{left-hand limit} \\ \text{of the CDF at } x}} = \underbrace{\lim_{y \uparrow x}}_{\substack{y \text{ approaches } \\ x \text{ from the left}}} \hspace{-2ex} F(y)$ 
2. $\mathbb{P}(x < X \leq y) = F(y) - F(x)$
3. $\mathbb{P}(X > x) = 1 - F(x)$
4. If $X$ is continuous then:

$$
\begin{align*}
F(b) - F(a) &= \mathbb{P}(a < X < b) = \mathbb{P}(a \leq X < b) \\
&= \mathbb{P}(a < X \leq b) = \mathbb{P}(a \leq X \leq b)
\end{align*}
$$

**Lemma**. If $X$ is continuous then $ \mathbb{P}(X = x) = 0 $ for every $x$.

**Proof**. A continuous distribution has no jumps (as they are continuous functions) and $F(x) - F(x^-)$ is defined as the jump size of the CDF at $x$. For a continuous variable:

$$
\begin{align*}
&F(x^-) = F(x) = F(x^+) \;\; \text{ for all } x \\
\therefore \quad &F(x) - F(x^-) = 0 \\
\implies &\mathbb{P}(X=x) = 0
\end{align*}
$$

Let $X$ be a random variable with CDF $F$. The **inverse CDF** or **quantile function** is defined as:

$$ F^{-1}(q) = \inf \{ x : F(x) > q \} \; \text{ for } q \in [0,1] $$

If $F$ is **strictly increasing** & **continuous** then $F^{-1}(q)$ yields a unique real number $x$ such that $F(x) = q$.

$ \ulcorner F^{-1}(\frac{1}{4}) : \text{ the first quartile} \quad \| \quad F^{-1}(\frac{1}{2}) : \text{ the median} \quad \| \quad F^{-1}(\frac{3}{4}) : \text{ the third quartile }\lrcorner $

$\inf$ **(infimum):** the greatest lower bound of a set<br>
$\phantom{\inf \text{(infimum):}}$ i.e. the largest value that is $\leq$ every element of the set

$\sup$ **(supremum):** the least upper bound of a set<br>
$\phantom{\sup \text{(supremum):}}$ i.e. the smallest value that is $\geq$ every element of the set

**Note:** The value generated from these do not have to be in the set.<br>
$\phantom{\text{Note:}}$ But if the $\sup / \inf$ are in the set, they are the maximum / minimum repsectively.

$\min$ **(minimum):** smallest element that actually belongs to the set

$\max$ **(maximum):** largest element that actually belongs to the set

Two random variables $X$ and $Y$ are equal in distribution, written $X \overset{d}{=} Y, $ if $F_X(x) = F_Y(x)$ for all $x$

### 2.3 Discrete Random Variables

$X \sim \text{F}$ is used to indicate that $X$ has distribution $\text{F}$.

#### Expectation and Variance

The **expected value** or **expectation** of a random variable is the theoretical mean of the random variable.<br>
It is the value representing the **mean** result if the variable were to be measured an infinite number of times.

$$
\begin{align*}
&\mu = \mathbb{E}[X] = \text{E}(X) = \sum x_i p_i \\
&\mu = \frac{\sum x}{n} \; \text{ or } \; \frac{\sum f x}{\sum f}
\end{align*}
$$

The **expectation** of a function $g(x)$ can be written as:

$$
\begin{align*}
X \text{ is discrete } : \mathbb{E} \left [ g(x) \right ] &= \sum_x g(x) p(x) \\
X \text{ is continuous } : \mathbb{E} \left [ g(x) \right ]  &= \int^{\infty}_{-\infty} g(x) f(x) \ dx
\end{align*}
$$

where:
- $p(x)$ is a **PMF** and $ \; p(x) = p(X=x)$
- $f(x)$ is a **PDF**

The **variance** of a random variable is the **expectation** of the squared distance of the random variable from the mean.<br>
It is the degree of **variation** that would be seen if the variable were to be repeatedly measured an infinite number of times.<br>
I.e. it is a measure of how **spread** out a variable is.

$$
\begin{align*}
&\sigma^2 = \mathbb{V}[X] = \text{Var}(X) &= \sum (x_i - \mu)^2 \cdot p_i &= \sum x_i^2 \cdot p_i - \mu^2  \\
&\phantom{x} &= \mathbb{E} \left [ \left (X - \mathbb{E}[X] \right )^2 \right ] &= \mathbb{E}\left [ X^2 \right ] - \mathbb{E}[X]^2 \\
&\sigma^2 = \frac{\sum (x - \mu)^2}{n} = \frac{\sum x^2}{n}-\mu^2 &\text{or }
\; \frac{\sum f(x - \mu)^2}{\sum f} &= \frac{\sum fx^2}{\sum f}-\mu^2
\end{align*}
$$

$\mathbb{E} \left [ (X - \mu )^2 \right ]: $ the expectation of the squared deviations from the mean<br>
$\sum (x_i - \mu)^2 \cdot p_i : $ the weighted average of the squared distances from the mean

where:
- **Distances from the mean:** $x_i - \mu$
- **Squared distances from the mean:** $(x_i - \mu)^2$
- **Weighted average:** each squared distance is weighted by its probability $p_i$

**Recall:**

$$
\begin{align*}
&\mu = \frac{\sum x}{n} \; \text{ or } \; \frac{\sum f x}{\sum f} \\
&\sigma^2 = \frac{\sum (x - \mu)^2}{n} = \frac{\sum x^2}{n}-\mu^2 \; \text{ or }
\; \frac{\sum f(x - \mu)^2}{\sum f} = \frac{\sum fx^2}{\sum f}-\mu^2
\end{align*}
$$

**Proof:**

$$
\begin{align*}
\frac{\sum (x - \mu)^2}{n} &= \frac{\sum x^2 - 2\mu x + \mu^2}{n} =
\frac{\sum x^2}{n} - \frac{2\mu \sum x}{n} + \frac{n \mu^2}{n} \\
&= \frac{\sum x^2}{n} - 2\mu^2 + \mu^2 = \frac{\sum x^2}{n} - \mu^2 \\ \\
\frac{\sum f(x - \mu)^2}{\sum f} &= \frac{\sum fx^2 - 2\mu f x + f\mu^2}{\sum f} =
\frac{\sum fx^2}{\sum f} - \frac{2\mu \sum fx}{\sum f} + \frac{\mu^2 \sum f}{\sum f} \\
&= \frac{\sum fx^2}{\sum f} - 2\mu^2 + \mu^2 = \frac{\sum fx^2}{\sum f} - \mu^2
\end{align*}
$$

#### Linear Transformations

| transformation |          mean          |       variance       |
|----------------|------------------------|----------------------|
| $a$ (constant) | $a$                    | $0$                  |
| $X$            | $\mathbb{E}[X]$        | $\text{Var}(X)$      |
| $aX + b$       | $a \mathbb{E}[X] + b$  | $a^2 \text{Var}(X)$  |
| $-X$           | $-\mathbb{E}[X]$       | $\text{Var}(X)$      |

As aforementioned $\sigma^2$ can also be written as: $\text{Var}(X) = \text{E}(X^2) - \text{E}(X)^2$<br>
Often read as: the mean of the squares $-$ the square of the means.

**Proof:**

$$
\begin{align*}
\text{Var}(X) &= \mathbb{E} \left [ (X - \text{E}(X))^2 \right ] \\
&= \mathbb{E} \left [ X^2 -2 \text{E}(X) \cdot X + \text{E}(X)^2 \right ] \\
&= \text{E}(X^2) + \text{E} \biggl ( \underbrace{-2 \text{E} (X)}_{\text{constant}} \cdot X \biggr ) +
\text{E} \biggl ( \underbrace{\text{E}(X)^2}_{\text{constant}} \biggr ) \\
&= \text{E}(X^2) -2 \text{E}(X) \cdot \text{E}(X) + \text{E}(X)^2 \\
&= \text{E}(X^2) -2 \text{E}(X)^2 + \text{E}(X)^2 \\
&= \text{E}(X^2) - \text{E}(X)^2
\end{align*}
$$

#### 1. The Point Mass Distribution (at $a$): $\quad \mu = a, \, \sigma^2 = 0$

$X$ has a point mass distribution at $a$ written $X \sim \delta_a$.<br>
If $\, \mathbb{P}(X=a) = 1$ then:

$$
F(x) =
\begin{cases}
0 &\text{if } x < a \\
1 &\text{if } x \geq a
\end{cases}
$$

The corresponding PMF is $f(x) = 1$ for $x = a \,$ & $\, 0$ otherwise.

#### 2. The Discrete Uniform Distribution: $\quad \mu = \frac{n+1}{2}, \, \sigma^2 = \frac{n^2-1}{12}$

Let $n>1$ be a given integer. If $X \sim \text{U}(n) \,$ then $X$ has a PMF given by:

$$
P(X=x) = f(x) =
\begin{cases}
\frac{1}{n} &\text{for } x = 1, \dotsc, n \\
0 &\text{otherwise}
\end{cases}
$$

**Note:** Any discrete uniform distribution distribution can be linearly transformed to the discrete uniform distribution of the whole numbers from $1$ to $n \; \left ( \text{U}(n) \right )$.

#### 3. The Bernoulli Distribution: $\quad \mu = p, \, \sigma^2 = pq$

Suppose we have a **single trial** that can result in one of **2 possible outcomes**, labelled success and failure.<br>
$\mathbb{P}(\text{Success}) = p \;\; \& \;\; \mathbb{P}(\text{Failure}) = 1 - p = q \,$ where $\, p \in [0,1]$. Let $X = 1$ if the trial results in a success $\, \& \; X = 0$ if the trial ends in failure. Then $X \sim \text{Bernoulli}(p):$

$$
P(X=x) = f(x) =
\begin{cases}
p^x(1-p)^{1-x} &\text{for } x \in [0,1] \\
0 &\text{otherwise}
\end{cases}
$$

#### 4. The Binomial Distribution: $\quad \mu = np, \, \sigma^2 = npq$

Suppose there are $\mathbf{n}$ **independent trials**. each trial can result in one of **2 possible outcomes** (success or failure). $\mathbb{P}(\text{Success}) = p \,$ (this **stays constant** from trial to trial), $\, \mathbb{P}(\text{Failure}) = 1 - p = q \,$ where $\, p \in [0,1]$. $X$ represents the number of successes in $n$ (independent bernoulli) trials. Then $X \sim \text{B}(n, p):$

$$
P(X=x) = f(x) =
\begin{cases}
\binom{n}{x} p^x(1-p)^{n-x} &\text{for } x = 0, \dotsc, n \\ \\
0 &\text{otherwise}
\end{cases}
$$

If $X_1 \sim \text{B}(n_1, p) \;\; \& \;\; X_2 \sim \text{B}(n_2, p)$<br>
Then $X_1 + X_2 \sim \text{B}(n_1 + n_2, p)$

**Terminology:**
- $X:$ a random variable
- $x:$ a particular value of the random variable
- $n, p:$ parameters

The parameter $p$ is usually unknown in an experiment & must be estimated from the data $\longrightarrow$ statistical inference.

A binomial distribution is used if we want to find the number of succcesses in $n$ independent bernoulli trials.

#### 5. The Geometric Distribution: $\quad \mu = \frac{1}{p}, \, \sigma^2 = \frac{q}{p^2}$

Suppose we have an an experiment with **independent trials**. Each trial results in one of **2 possible outcomes** (success or failure). $\mathbb{P}(\text{Success}) = p \,$ (this **stays constant** from trial to trial), $\, \mathbb{P}(\text{Failure}) = 1 - p = q \,$ where $\, p \in [0,1]$. $X$ represents the number of trials needed to get the first success. Then $X \sim \text{Geo}(p):$

$$
P(X=x) = f(x) = p(1-p)^{x-1} \;\; \text{ for } x \geq 1
$$

Additionally,

$$
\begin{align*}
\mathbb{P}(X>x) &= (1-p)^x \\
\text{The first success occurs after trial } x \,
&\iff \substack{\text{ No success occurs in the first } x \text{ trials } / \\
\text{ Started with } x \text{ consecutive failures}}
\end{align*}
$$

A geometric distribution is used if we want to find the number of trials needed to get the first success in an **infinite sequence** of independent and identically distributed bernoulli trials. We also have that:

$$
\begin{align*}
\sum_{x=1}^{\infty} \mathbb{P}(X=x) &= p \sum_{x=1}^{\infty} (1-p)^{x-1} \\
\text{ Let } m = x - 1 &\implies x = m + 1 \\
&= p \sum_{m=0}^{\infty} (1-p)^m \\
\text{Recall that when } &|r| < 1, \;\; S_\infty = \sum_{k=0}^{\infty} ar^k = \frac{a}{1-r} \\
&= p \cdot \frac{1}{1-(1-p)} \\
&= p \cdot \frac{1}{p} = 1
\end{align*}
$$

#### 6. The Poisson Distribution: $\quad \mu = \lambda, \, \sigma^2 = \lambda$

Suppose we have events that occur **randomly** and **independent** of each other.<br>
We also have that the average rate of events ($\lambda$) occuring must be **constant**. $X$ represents the number of events that occur in a fixed space or time interval. Then $X \sim \text{Po}(\lambda):$

$$
P(X=x) = f(x) = \frac{e^{-\lambda}\lambda^x}{x!} \;\; \text{ for } x \geq 0
$$

**Note that:** $\; \sum_{x=0}^{\infty} f(x) = 1$

**Proof:**

$$
\sum_{x=0}^{\infty} f(x) = \sum_{x=0}^{\infty} \frac{e^{-\lambda}\lambda^x}{x!} =
e^{-\lambda} \sum_{x=0}^{\infty} \frac{\lambda^x}{x!}
$$

**Sub-Proof:**

$$
\begin{align*}
\text{Recall the \textbf{Taylor series} expansion} &: \\
f(x) &= f(a) + f'(a) \frac{(x-a)}{1!}  + f^2(a) \frac{(x-a)^2}{2!} + \dotsb \\
&= \sum_{n=0}^{\infty} f^{(n)}(a) \frac{(x-a)^n}{n!} \\
\text{A \textbf{Maclaurin series} is a \textbf{Taylor series} where } a = 0 &: \\
f(x) &= \sum_{n=0}^{\infty} f^{(n)}(0) \frac{x^n}{n!} \\
\text{Let } f(x) = e^x &: \\
e^x &= 1 + \frac{x}{1!} + \frac{x^2}{2!} + \dotsb = \sum_{n=0}^{\infty} \frac{x^n}{n!}
\end{align*}
$$

**Proof Continued:**

$$
e^{-\lambda} \sum_{x=0}^{\infty} \frac{\lambda^x}{x!} = e^{-\lambda} e^\lambda = 1
$$

#### 7. The Hypergeometric Distribution: $\quad \mu = n\frac{A}{N}, \, \sigma^2 = n \frac{A}{N} \cdot \frac{N-A}{N} \cdot \frac{N-n}{N-1}$

Suppose we are randomly sampling $n$ objects **without replacement** from a source / population of $N$ objects. The result of each draw can be classified into one of **2 mutually exclusive categories** (success or failure). The source contains $A$ successes and $N-A$ failures. $X$ represents the number of successes in a sample.<br>
Then $X \sim \text{Hypergeometric}(N, A, n):$

$$
\begin{align*}
P(X=x) = f(a) &= \frac{\binom{A}{a} \binom{N-A}{n-a}}{\binom{N}{n}} \\
&\text{for } \max(0,\; n - (N-A)) \leq a \leq \min(n,\; A) \\ \\
\textbf{LHS} &: \max(0,\; n - (N-A)): \text{ minimum number of successes} \\
\textbf{RHS} &: \min(n,\; A): \text{ maximum number of successes}
\end{align*}
$$

where:
- $N:$ is the population size
- $A:$ is the number of successes in the population
- $n:$ is the number of draws / samples
- $a:$ is the number of observed successes
